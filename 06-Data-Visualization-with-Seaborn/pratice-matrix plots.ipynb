```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')
```

```python
tips.head()
```

```python
flights.head()
```

### HeatMap

```python
# how go with correlation data first
tc = tips.corr(numeric_only=True)
# now we have matrix form of some sort of data here.
# them we can call heatmap on it
```

```python
plt.figure(figsize=(3,3))
sns.heatmap(tc)
plt.show()
# Heat map really does is it colors in those values based on some sort of gradient scale.
# And that way if you have a very large correlation data frame or some other matrix form data you can
# easily tell what the numbers or what the values are relative to each other using a heat map.
# And that's the main primary use for a matrix plot.
```

```python
sns.heatmap(tc,annot=True) # it actually assign a actual value on it.
plt.show()
```

```python
sns.heatmap(tc,annot=True,cmap='coolwarm') # cmap -> actual color map we wants.
plt.show()
```

```python
# Now go with flights data.
# let create pivot table.
ft = flights.pivot(index='month',columns='year',values='passengers')
ft
```

```python
sns.heatmap(ft,cmap='magma',linewidths=1,linecolor='white') # And now I can see based on a color gradient the data in a lot of clear format.
plt.show()                                                  # linewidth, linecolor helps in seperate each boxes.
```

### Cluster Map

```python
sns.clustermap(ft)
plt.show()
```

```python
sns.clustermap(ft,cmap='coolwarm',standard_scale=1) 
plt.show()
```
