```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
iris = sns.load_dataset('iris') # dataset of bunch of different flower.
iris.head()
```

```python
iris['species'].unique()
```

```python
# pairplot is kind of automated joint plot for the entire dataset
# pairplot is the simplified version of PairGrid
sns.pairplot(iris)
plt.show()
```

```python
# In PairGrid we have a lot of customize ability apart from normal plot.
# It gives lot of control
sns.PairGrid(iris)
plt.show()
# it just return a emty pairgrid
```

```python
g = sns.PairGrid(iris) 
g.map(plt.scatter) # It assign scatter plot for diagonal,upper plot and lower plot
plt.show()
```

```python
# we can also specify each part with each plot.
g = sns.PairGrid(iris)
g.map_diag(sns.barplot)
g.map_upper(plt.scatter)
g.map_lower(sns.kdeplot)
plt.show()
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# FacetGrid -> here we need to pass column and row name.
g = sns.FacetGrid(data=tips,col='time',row='smoker')
plt.show() 
# it return empty grid
```

```python
# Now we map plot type to this grid.
g = sns.FacetGrid(data=tips,col='time',row='smoker') # here row with smoker type (yes, no) and column with time type (lunch, dinner)
g.map(sns.kdeplot,'total_bill') # x axes with total bill . kdeplot which wnat one variable
plt.show()
```

```python
g = sns.FacetGrid(data=tips,col='time',row='smoker') 
g.map(sns.scatterplot,'total_bill','tip') # x axes with total bill, y axes with tip. Here scatter plot need two variable.
plt.show()
```
