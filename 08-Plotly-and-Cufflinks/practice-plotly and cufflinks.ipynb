```python
import pandas as pd
import numpy as np
%matplotlib inline
```

```python
from plotly import __version__
from plotly.offline import download_plotlyjs, init_notebook_mode, plot, iplot
# These are the requirement we are taken from plotly 
# which supports in offline 
print(__version__)
```

```python
import cufflinks as cf
```

```python
init_notebook_mode(connected=True) 
# This is going to connect the javascript to your notebook
# Beacuse plotly essentially just connects pandas and pyhton to it intractive javascript library
# And this is going to allow your notebooks to access those visualizations.
```

```python
df = pd.DataFrame(np.random.randn(100,4),columns='A B C D'.split())
df.head()
```

```python
df2 = pd.DataFrame({'category':['a','b','c'], 'value':[34,2,90]})
df2
```

```python
df.plot() # Pandas has made automatically using matplotlib
```

```python
df.iplot() # now by using iplot -> the normal plot is converted into plotly interactive image
# By default we it use line plot.
```

```python
df.iplot(kind='scatter',x='A',y='B') 
# kind -> used for switch between plots.
# x,y are axes here we define the column name.
# And by default plotly is going to try to make a line to connect all these points.
```

```python
df.iplot(kind='scatter',x='A',y='B',mode='markers')# we can change the above by using mode.
```

```python
df2.head()
```

```python
# Lets go for bar plot
df2.iplot(kind='bar',x='category',y='value') # In this case it give some meaning because of dataset
```

```python
df.iplot(kind='bar')# But in this there is no proper understanding.
```

```python
df.sum().iplot(kind='bar') 
# But keep in mind that this bar plot with plotly becomes really powerful when you call some sort of aggregate
```

```python
# Boxplot automatically do that sort of aggregation for the distribution.
df.iplot(kind='box')
```

```python
# 3D surface plot
df3 = pd.DataFrame({'x':[1,2,4,5,9],'y':[34,23,90,89,12],'z':[100,665,903,564,778]})
df3
```

```python
df3.iplot(kind='surface')
```

```python
df4 = pd.DataFrame({'x':[1,2,3,4,5],'y':[34,23,90,89,12],'z':[5,4,3,2,1]})
df4
```

```python
df4.iplot(kind='surface',colorscale='rdylbu') # we can also change the color scale.
```

```python
df['A'].iplot(kind='hist',bins=50)# Plot for specific column ,we can also add more bins.
```

```python
df.iplot('hist') # It just returns overlapping histogram of all the columns.
# Meaning you can compare the histograms of columns and you car turn them on and off to compare them to each other
```

```python
# Spread type visualization are used a lot for stock data.
df[['A','B']].iplot(kind='spread') # it give line plot with spread which helps in analysis between two stocks.
```

```python
# Bubble plot is very similar to a scatterplot except that it will change size of the points based off of another variable.
# Ex in GDP in comparison to population,happiness factor etc.
df.iplot(kind='bubble',x='A',y='B',size='C')
```

```python
# scatter matrix plot is very similar to seaborn's pair plot.
df.scatter_matrix()
```
