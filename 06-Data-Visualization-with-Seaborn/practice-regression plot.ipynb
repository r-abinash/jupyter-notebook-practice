```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

### LMplot

```python
sns.lmplot(x='total_bill',y='tip',data=tips)# we want to pass x axes, y axes with data . it returns scatter plot with linear fit on top.
plt.show()
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex')
plt.show()# it return two scatter plot and linear fits for both male and female.
# To infer out of this plot is that male and female essentially have the same linear fit as far as total bill vs tip.
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex',markers=['o','v']) # here we can change the marker type for each hue. pass as list
plt.show() # These markers symbols come from the matplotlib marker symbols.
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex',markers=['o','v'],
          scatter_kws={'s':100}) # sns.lmplot is it's actually calling matplotlib very much under the hood.
plt.show() # If you want to effect matplotlib directly from seaborn.
# if we wnat to affect the scatter plot we need to pass it in dictionary.'s'->size.
```

```python
# instead of hue we have 'col' attribute to seperate the male and female chart
sns.lmplot(x='total_bill',y='tip',data=tips,col='sex')
plt.show() #So instead of separating it by color separated by columns
```

```python
# we can do the above row also
sns.lmplot(x='total_bill',y='tip',data=tips,col='sex',row='time')
plt.show()
# it is essentially do the frid call and fassett group.
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,col='time',hue='sex',aspect=)
plt.show()# aspect -> The ratio between the height and the width of each of these separate grids.
          # Size -> size of each grid.
```
