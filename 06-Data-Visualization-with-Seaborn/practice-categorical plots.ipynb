```python
import seaborn as sns
import matplotlib.pyplot as plt
# %matplotlib inline
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

## Bar plot

```python
sns.barplot(x='sex',y='total_bill',data=tips) # x-> categorical column, y -> numerical column.
# here bar plot shows the average or mean of total bill for categorical column.
plt.show()
```

```python
import numpy as np
```

```python
sns.barplot(x='sex',y='total_bill',data=tips,estimator=np.std)#by default the estimator is avg
plt.show() # This what the standard deviation of the total bill column is per category.
```

## count plot

```python
sns.countplot(x='sex',data=tips)
plt.show()
```

## Box plot and Violent plot

```python
sns.boxplot(x='day',y='total_bill',data=tips)# x-> categorical column, y-> numerical column
plt.show()
# This box plot shows the core tiles of the data set while the whiskers extend to show 
# the rest of the distribution except for these points that are outside of these whiskers.
# These are determined to be outliers.
```

```python
sns.boxplot(x='day',y='total_bill',data=tips,hue='smoker')
plt.show()
# hue-> that's going to allow us to kind of split up these box plots even further by another categorical column eg.smoker
# If I run this now I see the box plots split per day and also split by smoker status 
# And here you can now compare not only. 
# Day by day behavior for total bill but also by whether or not this person is a smoker.
```

```python
sns.violinplot(x='day',y='total_bill',data=tips)# violent plot give little more imformation.
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=tips,hue='sex')
plt.show()
```

## Strip plot

```python
sns.stripplot(x='day',y='total_bill',data=tips,jitter=False) # one issue is we can't tell how many points are overlap.so use 'jitter'
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,jitter=True)
# Jitter is going to add a little bit of random noise to separate some of these stacked points.
# And you can see the thicker and more points there are a lot more clear ver versus just the normal strip plot.
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,jitter=True,hue='sex') #here we use hue to find how many male & female comes .
```

```python
sns.swarmplot(x='day',y='total_bill',data=tips) 
# it is the combination of strip plot and violent plot
# not use in very large dataset, because it takes lots of computation to arrange,not able to scale them in graph.
```

```python
sns.violinplot(x='day',y='total_bill',data=tips)
sns.swarmplot(x='day',y='total_bill',data=tips,color='black')
# here every single point and corresponding KDE on top of it.
# it show lot of information 
# it more usefull for explore data analysis not for presentation.
```

```python
# sns.factorplot()
sns.factorplot(x='sex',y='total_bill',data=tips,kind='bar')
```
