```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.countplot(x='sex',data=tips)
plt.show()
```

```python
sns.set_style('darkgrid') # seaborn allows u to set stule for the chart by using 'set_style'
sns.countplot(x='sex',data=tips)
plt.show()
```

```python
sns.set_style('dark')
sns.countplot(x='sex',data=tips)
plt.show()
```

```python
sns.set_style('ticks')
sns.countplot(x='sex',data=tips)
plt.show()
```

```python
# we also able to remove the spine which is boundaries of the chart.by default it remove top,right.
sns.set_style('ticks')
sns.countplot(x='sex',data=tips)
sns.despine()
plt.show()
```

```python
# if we want to remove left and bottom means we can specify it in despine() function
sns.set_style('ticks')
sns.countplot(x='sex',data=tips)
sns.despine(left=True)
plt.show()
```

```python
# we can also change the figure size by using matplotlib.
# in this seaborn contact with matplotlib ,then it only can change the fig size.
plt.figure(figsize=(12,3)) # pass fig size in tuple.
sns.countplot(x='sex',data=tips)
plt.show()
```

```python
#we can also do the above thing by aspect .
sns.lmplot(x='total_bill',y='tip',aspect=4,data=tips)
plt.show()
```

```python
# then we do set the context in for seaborn by using set_context() -> like poster,note,paper,talk.by that the size of chart may varies.
sns.set_context('poster')
sns.countplot(x='sex',data=tips,palette='coolwarm')
plt.show()
```

```python
# we able to set the font size also.
sns.set_context('notebook',font_scale=4)
sns.countplot(x='sex',data=tips,palette='coolwarm')
plt.show()
```
