```python
df1_reset.plot.scatter(x='A',y='B',c='C',cmap='coolwarm')# c-> color, here based on clor clumn, the color will looks.
                                                         # here we get a 3D visuals by compare cloumn A,B,C
plt.show()
```

```python
df1_reset.plot.scatter(x='A',y='B',s=df1_reset['C'],cmap='coolwarm')
plt.show()
```

```python
df1_reset.plot.scatter(x='A',y='B',s=df1_reset['C']*100,cmap='coolwarm') #s-> size, here the size of A,B can be determined by C colum.
# for 's' we need to pass dataframe column.
plt.show()
```

```python
df = pd.DataFrame(np.random.randn(1000,2),columns=['a','b'])
df.head()
```

```python
df.plot.hexbin(x='a',y='b')
#Now this is essentially like a scatterplot except it's the hexagonal bins.
plt.show()
```

```python
df.plot.hexbin(x='a',y='b',gridsize=25,cmap='coolwarm')# the above grid size is very small we can change it 'gridsize' we can also change color map bu 'cmap'
plt.show()
```

```python
# kernal density estimation plot
df.plot.kde() # we can do it paricular column also column
plt.show()
```
